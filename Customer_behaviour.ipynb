# Creating column 'age_group'

```python
labels = ['Young-Adult','Adult','Middle-Aged','Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age','age_group']].head(10)
```

# Creating column 'purchase_frequency_days'

```python
# Creating a Dictonary where all catergorical values are given some numerical values
frequency_mapping = {
    'Fortnightly':14,
    'Monthly': 30,
    'Weekly': 7,
    'Quarterly': 90,
    'Bi-weekly': 14,
    'Annually': 365,
    'Every 3 months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)

```

```python
df[['frequency_of_purchases','purchase_frequency_days']].head(10)
```

```python
print(df['purchase_frequency_days'].isnull().sum())
```

```python
df['purchase_frequency_days'] = (
    df['purchase_frequency_days']
    .fillna(0)
    .astype(int)
)
```

```python
df['purchase_frequency_days'] = df['purchase_frequency_days'].astype(int)
```
